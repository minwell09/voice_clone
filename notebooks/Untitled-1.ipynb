### 2.1 Kien truc he thong Text To Speech
 
* NLP xu ly ngu am va ngu dieu cung voi nhip dieu va xuat no ra phien am cua van ban dau vao
* Digital Signal Process: Chuyen doi ban ghi ngu am nhan duoc thanh giong noi
### 2.2 Qua trinh xu ly
#### A. Text Analyzer
* Tien xu ly van ban: xoa bo nhung ky tu, tach doan, tach cau, lam sach, tich hop, chuyen doi va giam so chieu
* Phan tich co bon giai doan:
    - Phan doan tu vung
    - Phan loai tu
* Phan tich cu phap - ngu dieu xac dinh cau truc van ban lien quan chat che hon viec xac dinh ngu dieu
#### B. Letter to sound
* LTS la qua trinh chuyen doi van ban da duoc xu ly thanh chuoi phien am de he thong co the tao ra loi noi. Qua trinh phien am nay khong don thuan la tra cuu tu dien, vi cach phat am cua moi tu co the tuy vao ngu canh, bien the hinh thai va vi tri cua tu trong cau 
    

```python
ptest = 'HUSCHAUSC'
pattern = 'HUSC'
N = len(ptest)
total_count = 0
last_occurrence = -1  
for j in range(3, N):

    if ptest[j-3:j+1] == pattern:
        last_occurrence = j - 3
    

    if last_occurrence != -1:
        total_count += (last_occurrence + 1)

print(f"Tổng số chuỗi con thỏa mãn: {total_count}")
```
